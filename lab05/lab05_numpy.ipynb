```python
import pandas as pd
import numpy as np
# Series z listy — domyślny indeks (0, 1, 2, ...)
oceny = pd.Series([4.5, 3.0, 5.0, 3.5, 4.0])
print(oceny)
```

```python
pensje = pd.Series({
    'Anna': 5500,
    'Jan': 7200,
    'Ewa': 4800,
    'Marek': 9100,
    'Kasia': 6300
})

print(pensje)
```

```python
# 1. Kto zarabia najwięcej
print(pensje.idxmax())

# 2. Średnia pensja
print(pensje.mean())

# 3. Powyżej średniej
print(pensje[pensje > pensje.mean()])

# 4. Ile osób poniżej 6000
print((pensje < 6000).sum())
```

```python
# 1. Podwyżka 10%
pensje_10 = pensje * 1.10

# 2. Premia 500
pensje_500 = pensje + 500

# 3. Nowa pensja
nowe_pensje = (pensje * 1.10) + 500

# 4. Ranking
print(nowe_pensje.sort_values(ascending=False))
```

## Zad 3

```python
url = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/penguins.csv'
penguins = pd.read_csv(url)
```

```python
print(penguins.head(10))
print(penguins.shape)
print(penguins.info())
```

```python
print(penguins.describe())
print(penguins.describe(include='all'))
print(penguins['body_mass_g'].median())
```

```python
print(penguins['species'].value_counts())
print(penguins['island'].value_counts())
print(penguins['sex'].value_counts())
print(penguins['species'].value_counts(normalize=True))
```

```python
print(penguins.isna().sum())

missing_sex = penguins['sex'].isna().sum()
percent = missing_sex / len(penguins) * 100
print(percent)
```

```python
tips = pd.read_csv('https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv')
# Dodaj kolumnę z procentem napiwku
tips['tip_pct'] = (tips['tip'] / tips['total_bill'] * 100).round(1)
```

```python

print(tips['total_bill'].mean())

print(tips['tip_pct'].mean())

print(tips['day'].value_counts())

print(tips['smoker'].value_counts())

print(tips['tip'].describe())

print(tips['time'].value_counts(normalize=True))

print(tips['size'].nunique())

```
